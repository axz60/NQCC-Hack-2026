Rolls-Royce PLC designs and manufactures power systems for a wide range of applications, including aviation engines and marine propulsion systems. We are also developing next-generation energy technologies, including hydrogen combustion systems and small modular reactors, while continuing to deliver the efficiency, reliability, and engineering excellence for which Rolls-Royce is known.

Computational simulation is fundamental to our development process. Accurate modelling of fluid flow, combustion, materials, structures, and component lifetimes enables informed, safety-critical engineering decisions. Our activities range from machine-learning-assisted design optimisation to large-scale GPU-accelerated simulations.

Many of these workloads involve structured mathematical models and large-scale linear algebra. As system complexity increases, the computational cost of higher-fidelity simulations continues to grow. Quantum computing offers a potential pathway to accelerate selected simulation and optimisation workloads.

Rolls-Royce has already explored quantum linear solvers for computational fluid dynamics, successfully emulating these methods on AMD GPU platforms. Building on this work, we are now investigating quantum chemistry and materials modelling for hydrogen technologies, protective coatings, and advanced energy systems.

This hackathon project forms part of our broader effort to evaluate quantum computing workflows and develop internal expertise in quantum simulation for future industrial applications.



## 3. Please provide a general description of the problem your use case aims to address and why this is important for your organisation

The transition towards sustainable fuels, including hydrogen-based propulsion and energy systems, introduces new materials challenges for aerospace and power-generation applications. In particular, hydrogen can interact with structural materials and protective coatings, leading to degradation mechanisms that may impact reliability, performance, and component lifetime. Developing materials that can withstand these environments is therefore an important challenge for Rolls-Royce as we investigate future propulsion and energy technologies.

A key requirement for understanding material behaviour is the ability to accurately model chemical reactions and electronic structure from first principles. Quantities such as ground-state energies, reaction pathways, and electronic properties provide valuable insight into material stability and performance. However, these calculations can become increasingly computationally demanding for larger and more realistic systems.

This project explores whether quantum computing methods can contribute to future materials-modelling workflows. Specifically, we will develop and benchmark a NISQ-compatible approach based on Variational Quantum Eigensolvers (VQE) and Hamiltonian moments. We will investigate whether low-order Hamiltonian moments can reproduce ground-state energies through the Galitskii-Migdal formalism, providing a foundation for future Green's function and spectral-property calculations.

The methodology will be validated against classical CCSD calculations for the hydrogen molecule and hydrogen-exchange reactions before being applied to a hydrogen interaction with boron nitride, a material relevant to protective coatings for hydrogen technologies. By comparing classical emulators and quantum hardware, we aim to assess both the capabilities and current limitations of quantum computing for industrially relevant quantum chemistry problems.


## 4. Technical Description

This use case falls within the domain of quantum simulation and electronic-structure modelling. The long-term objective is to develop quantum computing workflows that can support the study of materials relevant to sustainable energy technologies, including hydrogen-powered propulsion systems and associated protective coatings.

Current electronic-structure calculations are typically performed using first-principles quantum chemistry methods such as Density Functional Theory (DFT), Coupled Cluster (CCSD), and Full Configuration Interaction (FCI). These approaches are widely used because they provide accurate predictions of molecular energies and reaction pathways. However, their computational cost increases rapidly with system size. FCI scales exponentially with the number of orbitals, while high-accuracy coupled-cluster methods exhibit steep polynomial scaling. Consequently, simulations of strongly correlated systems and realistic materials can become computationally demanding even on modern HPC platforms.

The problem may be formulated through the second-quantized electronic Hamiltonian

H = Σᵢⱼ hᵢⱼ a†ᵢaⱼ + ½ Σᵢⱼₖₗ hᵢⱼₖₗ a†ᵢa†ⱼaₗaₖ,

where hᵢⱼ and hᵢⱼₖₗ are one- and two-electron integrals obtained from PySCF. Following a fermion-to-qubit mapping, such as the Jordan-Wigner transformation, the Hamiltonian is represented as a weighted sum of Pauli operators,

H = Σᵢ cᵢPᵢ.

The proposed quantum workflow consists of two stages. First, a Variational Quantum Eigensolver (VQE) will be used to prepare an approximate ground state |Ψ₀⟩ and estimate the ground-state energy E₀ = ⟨Ψ₀|H|Ψ₀⟩.

Second, Hamiltonian moments will be calculated from the prepared quantum state. These quantities are defined as

μₙ = ⟨Ψ₀|Hⁿ|Ψ₀⟩,

and provide a compact description of the energy distribution associated with a quantum state. Hamiltonian moments are closely related to Green's functions and spectral properties and can be used to construct systematic approximations to spectral functions and other electronic observables. During the hackathon we will focus on the zeroth- and first-order moments as an initial proof-of-concept, while developing a workflow that can be extended to higher-order moments and Green's function calculations in future work.

The methodology will be benchmarked against classical CCSD calculations for the hydrogen molecule and hydrogen-exchange reactions. Following validation, the workflow will be applied to a hydrogen interaction with boron nitride using an active-space Hamiltonian. Calculations will be performed using both statevector simulators and available quantum hardware.

Success criteria include: (i) agreement between VQE and moment-derived energy estimates, (ii) agreement with CCSD reference calculations, (iii) successful execution of the workflow on quantum hardware, and (iv) quantitative assessment of the impact of noise and finite-shot sampling on Hamiltonian-moment calculations.


## 5. Data

This use case does not rely on proprietary or personal data. All systems studied during the hackathon will be generated from first-principles quantum chemistry calculations and publicly available molecular structures.

The primary data will consist of molecular geometries, one-electron integrals, two-electron integrals, and active-space Hamiltonians generated using the PySCF quantum chemistry package. Small benchmark systems, including H₂ and hydrogen-exchange reactions, will be used to validate the methodology. Following validation, an active-space model of a hydrogen interaction with boron nitride will be constructed to investigate a materials-relevant application.

To reduce the computational complexity of the boron nitride system, Complete Active Space Self-Consistent Field (CASSCF) calculations will be used to identify a chemically relevant active space. The resulting effective Hamiltonian will then be mapped onto a qubit representation for use in the quantum algorithms.

All data used in the project will be synthetic and generated computationally. No personal, confidential, export-controlled, or GDPR-sensitive data will be used.


## Impact 

Quantum computing is a rapidly developing field, but many industrial applications remain at an early stage of maturity. While optimisation problems have demonstrated promising results on NISQ-era hardware, simulation workflows for materials science require further development before they can be deployed in engineering environments.

This project focuses on developing quantum simulation methods relevant to hydrogen-powered technologies. For Rolls-Royce, the challenge is not only to identify new materials, but also to understand how existing materials and protective coatings behave in hydrogen-rich environments and how they may be protected from degradation.

The use case aligns with UN Sustainable Development Goal 7 (Affordable and Clean Energy) through its relevance to hydrogen energy systems, and Sustainable Development Goal 9 (Industry, Innovation and Infrastructure) through the development of advanced computational tools for engineering and materials science.

From a responsible innovation perspective, the project seeks to provide an evidence-based assessment of where quantum computing can realistically contribute to industrial challenges. The work uses synthetic scientific data and presents minimal ethical risk. More broadly, it reflects the principle that meaningful societal impact often arises through incremental improvements to the tools and infrastructure that enable future technologies, rather than through isolated technological breakthroughs alone.
